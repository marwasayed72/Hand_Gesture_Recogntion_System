```python
import os
import numpy as np
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import load_model


UNSEEN_DIR = "/kaggle/working/asl_alphabet_unseen_sample"
MODEL_PATH = "/kaggle/input/mobile/keras/default/1/asl_model.h5" 

model = load_model(MODEL_PATH)
print("✅ Model loaded successfully!")


datagen = ImageDataGenerator(rescale=1./255)


unseen_generator = datagen.flow_from_directory(
    UNSEEN_DIR,
    target_size=(128, 128),
    color_mode="rgb",
    class_mode="categorical",
    shuffle=False
)


print("\n🚀 Evaluating model on unseen dataset...")
loss, acc = model.evaluate(unseen_generator, verbose=1)
print(f"\n📊 Unseen Test Accuracy: {acc*100:.2f}%")
```

```python
import os
import numpy as np
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import load_model


UNSEEN_DIR = "/kaggle/working/asl_alphabet_unseen_sample"
MODEL_PATH = "/kaggle/input/cnn/keras/default/1/best.h5"


model2 = load_model(MODEL_PATH)
print("✅ Model loaded successfully!")


datagen = ImageDataGenerator(rescale=1./255)

unseen_generator = datagen.flow_from_directory(
    UNSEEN_DIR,
    target_size=(64, 64),      
    color_mode="grayscale",    
    class_mode="categorical",
    shuffle=False
)


print("\n🚀 Evaluating model on unseen dataset...")
loss, acc = model2.evaluate(unseen_generator, verbose=1)
print(f"\n📊 Unseen Test Accuracy: {acc*100:.2f}%")
print(f"📉 Unseen Test Loss: {loss:.4f}")
```
