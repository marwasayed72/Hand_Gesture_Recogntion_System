```python
import random, shutil,os ,glob
```

```python
print(">> Available datasets under /kaggle/input:")
for d in os.listdir("/kaggle/input"):
    print(" -", d)
```

```python
TRAIN_DIR = "/kaggle/input/aslamerican-sign-language-aplhabet-dataset/ASL_Alphabet_Dataset/asl_alphabet_train"
```

```python
random.seed(42) 
SELECTED_CLASSES = None 
N_PER_CLASS = 1000
SRC_DIR = TRAIN_DIR
SAMPLE_DIR = "/kaggle/working/asl_sample_train1"
os.makedirs(SAMPLE_DIR, exist_ok=True)
IMG_EXTS = {".jpg", ".jpeg", ".png", ".bmp"}
all_classes = [d for d in os.listdir(SRC_DIR) if os.path.isdir(os.path.join(SRC_DIR, d))]
classes = SELECTED_CLASSES if SELECTED_CLASSES else sorted(all_classes)
print(f"Working on {len(classes)} classes")
for cls in classes:
    src_cls = os.path.join(SRC_DIR, cls)
    dst_cls = os.path.join(SAMPLE_DIR, cls)
    os.makedirs(dst_cls, exist_ok=True)
    files = [f for f in os.listdir(src_cls) if os.path.splitext(f)[1].lower() in IMG_EXTS]
    chosen = random.sample(files, min(N_PER_CLASS, len(files)))
    for f in chosen:
        shutil.copy(os.path.join(src_cls, f), os.path.join(dst_cls, f))
    


```
